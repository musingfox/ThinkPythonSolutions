# 類別和方法

Python 是一種**物件導向語言**——也就是說，它提供支援物件導向程式設計的功能，具有以下定義特徵：

-   大部分計算以物件操作的形式表達。

-   物件通常代表現實世界中的事物，而方法通常對應於現實世界中事物互動的方式。

-   程式包含類別和方法定義。

例如，在前一章中，我們定義了一個 `Time` 類別，對應於人們記錄一天中時間的方式，我們定義了對應於人們對時間進行操作的函數。
但在 `Time` 類別的定義與其後的函數定義之間沒有明確的連接。
我們可以通過將函數重寫為**方法**來使連接明確，方法是在類別定義內部定義的。

## 定義方法

在前一章中，我們定義了一個名為 `Time` 的類別，並寫了一個名為 `print_time` 的函數來顯示一天中的時間。

```python
class Time:
    """Represents the time of day."""

def print_time(time):
    s = f'{time.hour:02d}:{time.minute:02d}:{time.second:02d}'
    print(s)
```

要使 `print_time` 成為方法，我們只需要將函數
定義移到類別定義內部。注意縮排的變化。

同時，我們將參數的名稱從 `time` 改為 `self`。
這個改變不是必需的，但方法的第一個參數命名為 `self` 是慣例。

```python
class Time:
    """Represents the time of day."""    

    def print_time(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        print(s)
```

要呼叫此方法，你必須傳遞一個 `Time` 物件作為參數。
以下是我們用來製作 `Time` 物件的函數。

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

以下是一個 `Time` 實例。

```python
start = make_time(9, 40, 0)
```

現在有兩種方式呼叫 `print_time`。第一種（較不常見的）
方式是使用函數語法。

```python
Time.print_time(start)
```

在這個版本中，`Time` 是類別的名稱，`print_time` 是方法的名稱，`start` 作為參數傳遞。
第二種（更慣用的）方式是使用方法語法：

```python
start.print_time()
```

在這個版本中，`start` 是方法被調用的物件，稱為**接收者**，基於調用方法就像向物件發送訊息的比喻。

無論語法如何，方法的行為都是相同的。
接收者被指派給第一個參數，所以在方法內部，`self` 指向與 `start` 相同的物件。

## 另一個方法

以下是前一章中的 `time_to_int` 函數。

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

以下是重寫為方法的版本。

```python
%%add_method_to Time

    def time_to_int(self):
        minutes = self.hour * 60 + self.minute
        seconds = minutes * 60 + self.second
        return seconds
```

第一行使用特殊命令 `add_method_to`，它將方法添加到先前定義的類別中。
此命令在 Jupyter notebook 中有效，但它不是 Python 的一部分，所以在其他環境中不會運作。
通常，類別的所有方法都在類別定義內部，所以它們與類別同時定義。
但對於本書來說，一次定義一個方法很有幫助。

如前面的範例一樣，方法定義有縮排，參數名稱是 `self`。
除此之外，方法與函數相同。
以下是我們如何調用它。

```python
start.time_to_int()
```

我們通常說「呼叫」函數和「調用」方法，但它們意思相同。

## 靜態方法

作為另一個範例，讓我們考慮 `int_to_time` 函數。
以下是前一章的版本。

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

此函數接受 `seconds` 作為參數並返回新的 `Time` 物件。
如果我們將其轉換為 `Time` 類別的方法，我們必須在 `Time` 物件上調用它。
但如果我們試圖創建新的 `Time` 物件，我們應該在什麼上調用它？

我們可以使用**靜態方法**來解決這個雞生蛋、蛋生雞的問題，靜態方法是不需要類別實例就能調用的方法。
以下是我們如何將此函數重寫為靜態方法。

```python
%%add_method_to Time

    def int_to_time(seconds):
        minute, second = divmod(seconds, 60)
        hour, minute = divmod(minute, 60)
        return make_time(hour, minute, second)
```

因為它是靜態方法，所以沒有 `self` 作為參數。
要調用它，我們使用 `Time`，它是類別物件。

```python
start = Time.int_to_time(34800)
```

結果是一個代表 9:40 的新物件。

```python
start.print_time()
```

現在我們有了 `Time.from_seconds`，我們可以用它來將 `add_time` 寫成方法。
以下是前一章的函數。

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

以下是重寫為方法的版本。

```python
%%add_method_to Time

    def add_time(self, hours, minutes, seconds):
        duration = make_time(hours, minutes, seconds)
        seconds = time_to_int(self) + time_to_int(duration)
        return Time.int_to_time(seconds)
```

`add_time` 有 `self` 作為參數，因為它不是靜態方法。
它是普通方法——也稱為**實例方法**。
要調用它，我們需要一個 `Time` 實例。

```python
end = start.add_time(1, 32, 0)
print_time(end)
```

## 比較 Time 物件

作為另一個範例，讓我們將 `is_after` 寫成方法。
以下是 `is_after` 函數，它是前一章練習的解答。

```python
def is_after(t1, t2):
    return time_to_int(t1) > time_to_int(t2)
```

以下是作為方法的版本。

```python
%%add_method_to Time

    def is_after(self, other):
        return self.time_to_int() > other.time_to_int()
```

因為我們要比較兩個物件，第一個參數是 `self`，我們將第二個參數稱為 `other`。
要使用此方法，我們必須在一個物件上調用它並將
另一個作為參數傳遞。

```python
end.is_after(start)
```

這種語法的一個好處是它幾乎讀起來像一個問題，
「`end` 在 `start` 之後嗎？」

## `__str__` 方法

當你寫方法時，你可以選擇幾乎任何你想要的名稱。
但是，有些名稱有特殊含義。
例如，如果物件有名為 `__str__` 的方法，Python 會使用該方法將物件轉換為字串。
例如，以下是時間物件的 `__str__` 方法。

```python
%%add_method_to Time

    def __str__(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        return s
```

此方法類似於前一章的 `print_time`，不同之處在於它返回字串而不是印出它。

你可以用通常的方式調用此方法。

```python
end.__str__()
```

但 Python 也可以為你調用它。
如果你使用內建函數 `str` 將 `Time` 物件轉換為字串，Python 會使用 `Time` 類別中的 `__str__` 方法。

```python
str(end)
```

如果你印出 `Time` 物件，它也會做同樣的事。

```python
print(end)
```

像 `__str__` 這樣的方法稱為**特殊方法**。
你可以通過它們的名稱以兩個下底線開始和結束來識別它們。

## __init__ 方法

特殊方法中最特殊的是 `__init__`，之所以這樣稱呼是因為它初始化新物件的屬性。
`Time` 類別的 `__init__` 方法可能看起來像這樣：

```python
%%add_method_to Time

    def __init__(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second
```

現在當我們實例化 `Time` 物件時，Python 會調用 `__init__`，並傳遞參數。
所以我們可以同時創建物件並初始化屬性。

```python
time = Time(9, 40, 0)
print(time)
```

在此範例中，參數是可選的，所以如果你不帶參數呼叫 `Time`，
你會得到預設值。

```python
time = Time()
print(time)
```

如果你提供一個參數，它會覆蓋 `hour`：

```python
time = Time(9)
print(time)
```

如果你提供兩個參數，它們會覆蓋 `hour` 和 `minute`。

```python
time = Time(9, 45)
print(time)
```

如果你提供三個參數，它們會覆蓋所有三個預設
值。

當我寫新類別時，我幾乎總是從寫 `__init__` 開始，這使得創建物件更容易，以及 `__str__`，這對除錯很有用。

## 運算子重載

通過定義其他特殊方法，你可以指定
程式設計師定義類型上運算子的行為。例如，如果你為 `Time` 類別定義一個名為 `__add__` 的方法，你可以在 Time 物件上使用 `+` 運算子。

以下是 `__add__` 方法。

```python
%%add_method_to Time

    def __add__(self, other):
        seconds = self.time_to_int() + other.time_to_int()
        return Time.int_to_time(seconds)
```

我們可以像這樣使用它。

```python
duration = Time(1, 32)
end = start + duration
print(end)
```

當我們運行這三行程式碼時發生了很多事情：

* 當我們實例化 `Time` 物件時，`__init__` 方法被調用。

* 當我們在 `Time` 物件上使用 `+` 運算子時，其 `__add__` 方法被調用。

* 當我們印出 `Time` 物件時，其 `__str__` 方法被調用。

改變運算子的行為使其能與程式設計師定義的類型一起運作稱為**運算子重載**。
對於每個運算子，如 `+`，都有對應的特殊方法，如 `__add__`。

## 除錯

如果 `minute` 和 `second` 的值在 `0` 和 `60` 之間（包括 `0` 但不包括 `60`），並且如果 `hour` 是正數，則 `Time` 物件是有效的。
此外，`hour` 和 `minute` 應該是整數值，但我們可能允許 `second` 有小數部分。
像這樣的要求稱為**不變量**，因為它們應該始終為真。
換句話說，如果它們不為真，就是出了什麼問題。

寫程式碼來檢查不變量可以幫助檢測錯誤並找到其原因。
例如，你可能有一個像 `is_valid` 這樣的方法，它接受 Time 物件，如果它違反不變量則返回 `False`。

```python
%%add_method_to Time

    def is_valid(self):
        if self.hour < 0 or self.minute < 0 or self.second < 0:
            return False
        if self.minute >= 60 or self.second >= 60:
            return False
        if not isinstance(self.hour, int):
            return False
        if not isinstance(self.minute, int):
            return False
        return True
```

然後，在每個方法的開始，你可以檢查參數以確保它們有效。

```python
%%add_method_to Time

    def is_after(self, other):
        assert self.is_valid(), 'self is not a valid Time'
        assert other.is_valid(), 'self is not a valid Time'
        return self.time_to_int() > other.time_to_int()
```

`assert` 陳述式評估後面的表達式。如果結果是 `True`，它不做任何事；如果結果是 `False`，它會導致 `AssertionError`。
以下是一個範例。

```python
duration = Time(minute=132)
print(duration)
```

```python
%%expect AssertionError

start.is_after(duration)
```

`assert` 陳述式很有用，因為它們區分處理正常條件的程式碼和檢查錯誤的程式碼。

## 詞彙表

**物件導向語言：**
提供支援物件導向程式設計功能的語言，特別是使用者定義的類型。

**方法：**
在類別定義內部定義並在該類別實例上調用的函數。

**接收者：**
方法被調用的物件。

**靜態方法：**
可以不需要物件作為接收者就能調用的方法。

**實例方法：**
必須以物件作為接收者調用的方法。

**特殊方法：**
改變運算子和某些函數與物件工作方式的方法。

**運算子重載：**
使用特殊方法改變運算子與使用者定義類型工作方式的過程。

**不變量：**
在程式執行期間應該始終為真的條件。

### 練習

在前一章中，一系列練習要求你寫一個 `Date` 類別和幾個與 `Date` 物件一起運作的函數。
現在讓我們練習將那些函數重寫為方法。

1. 寫一個 `Date` 類別的定義來代表日期——即年、月和日。

2. 寫一個 `__init__` 方法，它接受 `year`、`month` 和 `day` 作為參數並將參數指派給屬性。創建一個代表 1933 年 6 月 22 日的物件。

3. 寫 `__str__` 方法，它使用 f-字串來格式化屬性並返回結果。如果你用你創建的 `Date` 測試它，結果應該是 `1933-06-22`。

4. 寫一個名為 `is_after` 的方法，它接受兩個 `Date` 物件，如果第一個在第二個之後則返回 `True`。創建一個代表 1933 年 9 月 17 日的第二個物件，並檢查它是否在第一個物件之後。

提示：你可能會發現寫一個名為 `to_tuple` 的方法很有用，它返回一個包含 `Date` 物件屬性按年-月-日順序的元組。

```python
# Solution

class Date:
    
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    def __str__(date):
        s = f'{date.year}-{date.month:02d}-{date.day:02d}'
        return s

    def to_tuple(self):
        return self.year, self.month, self.day

    def is_after(self, other):
        t1 = self.to_tuple()
        t2 = other.to_tuple()
        return t1 > t2
```

你可以使用這些範例來測試你的解答。

```python
birthday1 = Date(1933, 6, 22)
print(birthday1)
```

```python
birthday2 = Date(1933, 9, 17)
print(birthday2)
```

```python
birthday1.is_after(birthday2)  # should be False
```

```python
birthday2.is_after(birthday1)  # should be True
```
