# 前言

## 這本書適合誰？

如果你想學程式設計，你來對地方了。

Python 是最適合初學者的程式語言之一，而且也是目前最受歡迎的技能之一。

你來的時間也很剛好，因為現在學程式可能比以往更容易。

有了像 ChatGPT 這樣的虛擬助理，你不必孤軍奮戰。

在這本書中，我會建議你如何利用這些工具加速學習。

這本書主要是為從未寫過程式的人，以及有其他程式語言經驗的人而寫。

如果你已經有相當的 Python 經驗，可能會覺得前幾章進度太慢。

學程式的一個挑戰是你得學會 *兩種* 語言：一種是程式語言本身；另一種是我們用來談論程式的專有詞彙。

如果你只學程式語言，當你需要解讀錯誤訊息、閱讀文件、與他人溝通或使用虛擬助理時，可能會遇到困難。

如果你有寫過程式，但還沒學會第二種語言，我希望這本書能幫助你。

## 本書目標

寫這本書時，我盡量謹慎使用詞彙。

每個術語在第一次出現時都會定義。

每章末尾都有一個詞彙表，回顧當章介紹的術語。

我也盡量簡潔。

讀這本書花的心力越少，你就有越多精力寫程式。

但你不能只靠讀書學程式——你必須練習。

因此，每章末都有練習題，讓你實際操作所學。

如果你認真閱讀並持續練習，你會有進步。

但我要提醒你，學程式不容易，即使是有經驗的程式設計師也會感到挫折。

我會建議一些策略，幫助你寫出正確的程式，並修正錯誤。

## 如何使用本書

本書的每一章都建立在前一章的基礎上，所以你應該依序閱讀，並花時間完成練習再繼續下一章。

前六章介紹基本元素，如算術、條件判斷和迴圈。

也介紹程式設計中最重要的概念：函式，以及一種強大的用法：遞迴。

第七、八章介紹字串——可以代表字母、單字和句子——以及處理字串的演算法。

第九到十二章介紹 Python 的核心資料結構——串列、字典和元組——這些是寫高效程式的強大工具。

第十二章介紹分析文字和隨機產生新文字的演算法。

這類演算法是大型語言模型（LLM）的核心，因此這章會讓你了解像 ChatGPT 這類工具的運作原理。

第十三章談如何將資料存到長期儲存裝置——檔案和資料庫。

練習題是寫一個程式，搜尋檔案系統並找出重複檔案。

第十四到十七章介紹物件導向程式設計（OOP），這是一種組織程式和資料的方式。

許多 Python 函式庫都是用物件導向風格寫的，這些章節會幫你理解它們的設計，並定義自己的物件。

本書的目標不是涵蓋整個 Python 語言。

我專注於一個子集，提供最強大的功能，且概念最少。

不過，Python 有很多功能可以有效解決常見問題。

第十八章介紹其中一些功能。

最後，第十九章是我的結語和繼續程式設計之路的建議。

## 第三版的新內容

這版最大的改變來自兩項新技術——Jupyter 筆記本和虛擬助理。

本書每章都是一個 Jupyter 筆記本，這是一種同時包含一般文字和程式碼的文件。

對我來說，這讓寫程式碼、測試和保持文字一致更容易。

對你來說，代表你可以在同一個地方執行程式碼、修改它、做練習。

第一章有使用筆記本的說明。

另一個大改變是我加入了使用虛擬助理（像 ChatGPT）加速學習的建議。

2016 年本書前一版出版時，這些工具的前身還不太實用，大多數人也不曉得它們。

現在它們是軟體工程的標準工具，我認為它們會是學程式和學很多其他東西的革命性工具。

書中其他改變是我對第二版的遺憾所致。

第一是我沒強調軟體測試。

2016 年已經是遺憾了，但有了虛擬助理，自動化測試更重要。

所以這版介紹 Python 最常用的測試工具 `doctest` 和 `unittest`，並包含多個練習題讓你練習。

另一個遺憾是第二版的練習題不均衡——有些有趣，有些不然，有些太難。

改用 Jupyter 筆記本幫我開發和測試更吸引人且有效的練習序列。

這版的主題順序幾乎沒變，但我調整了幾章的順序，並把兩個短章合併。

也擴充了字串的涵蓋範圍，包含正規表達式。

有幾章用到烏龜繪圖。

前幾版用 Python 的 `turtle` 模組，但它在 Jupyter 筆記本無法運作。

所以我換成一個新的 `jupyturtle` 模組，應該比較好用。

最後，我重寫了不少文字，釐清需要釐清的地方，刪減不夠簡潔的部分。

我對這版很驕傲——希望你喜歡！

## 開始使用

大多數程式語言（包含 Python）都有很多工具可以用來寫和執行程式。

這些工具叫做整合開發環境（IDE）。

一般來說，有兩種 IDE：

* 一種是用來編輯和執行程式碼檔案的工具。

* 另一種主要是用筆記本的工具，筆記本是同時包含文字和程式碼的文件。

對初學者，我建議從像 Jupyter 這樣的筆記本開發環境開始。

本書的筆記本可以從線上資源庫取得，網址是 <https://allendowney.github.io/ThinkPython>。

有兩種使用方式：

* 你可以下載筆記本，在自己的電腦執行。這樣你得安裝 Python 和 Jupyter，雖然不難，但如果你想學 Python，花很多時間安裝軟體可能會讓你挫折。

* 另一種是用 Colab 執行筆記本，Colab 是一個在瀏覽器執行的 Jupyter 環境，不用安裝任何東西。Colab 是 Google 提供的，免費使用。

如果你剛開始，我強烈建議你用 Colab。
