# 元組

本章介紹另一種內建型別——元組，然後展示串列、字典和元組如何協同工作。
它還介紹了元組指派以及函數可變長度參數串列的有用功能：封裝和解封裝運算子。

在練習中，我們將使用元組、串列和字典來解決更多的單詞謎題並實作高效的演算法。

注意：「tuple」有兩種發音方式。
有些人說「tuh-ple」，與「supple」押韻。
但在程式設計的語境中，大多數人說「too-ple」，與「quadruple」押韻。

## 元組就像串列

元組是一個值的序列。值可以是任何型別，它們由整數索引，所以元組很像串列。
重要的差別是元組是不可變的。

要建立元組，你可以寫一個以逗號分隔的值串列。

```python
t = 'l', 'u', 'p', 'i', 'n'
type(t)
```

雖然不是必需的，但通常會用括號將元組包圍起來。

```python
t = ('l', 'u', 'p', 'i', 'n')
type(t)
```

要建立只有一個元素的元組，你必須包含最後的逗號。

```python
t1 = 'p',
type(t1)
```

括號中的單一值不是元組。

```python
t2 = ('p')
type(t2)
```

建立元組的另一種方法是使用內建函數 `tuple`。沒有參數時，它建立一個空元組。

```python
t = tuple()
t
```

如果參數是序列（字串、串列或元組），結果是一個包含序列元素的元組。

```python
t = tuple('lupin')
t
```

因為 `tuple` 是內建函數的名稱，你應該避免將它用作變數名稱。

大多數串列運算子也適用於元組。
例如，方括號運算子索引一個元素。

```python
t[0]
```

切片運算子選擇一個元素範圍。

```python
t[1:3]
```

`+` 運算子連接元組。

```python
tuple('lup') + ('i', 'n')
```

`*` 運算子將元組複製指定的次數。

```python
tuple('spam') * 2 
```

`sorted` 函數適用於元組——但結果是串列，不是元組。

```python
sorted(t)
```

`reversed` 函數也適用於元組。

```python
reversed(t)
```

結果是一個 `reversed` 物件，我們可以將它轉換為串列或元組。

```python
tuple(reversed(t))
```

基於到目前為止的例子，元組看起來可能與串列相同。

## 但元組是不可變的

如果你嘗試用方括號運算子修改元組，你會得到 `TypeError`。

```python
%%expect TypeError
t[0] = 'L'
```

元組沒有任何修改串列的方法，如 `append` 和 `remove`。

```python
%%expect AttributeError

t.remove('l')
```

回想「屬性」是與物件關聯的變數或方法——這個錯誤訊息意味著元組沒有名為 `remove` 的方法。

因為元組是不可變的，它們是可雜湊的，這意味著它們可以用作字典中的鍵。
例如，以下字典包含兩個元組作為鍵，對應到整數。

```python
d = {}
d[1, 2] = 3
d[3, 4] = 7
```

我們可以像這樣在字典中查找元組：

```python
d[1, 2]
```

或者如果我們有一個引用元組的變數，我們可以將它用作鍵。

```python
t = (3, 4)
d[t]
```

元組也可以作為字典中的值出現。

```python
t = tuple('abc')
d = {'key': t}
d
```

## 元組指派

你可以將變數的元組放在指派的左側，將值的元組放在右側。

```python
a, b = 1, 2
```

值從左到右指派給變數——在這個例子中，`a` 得到值 `1`，`b` 得到值 `2`。
我們可以像這樣顯示結果：

```python
a, b
```

更一般地說，如果指派的左側是元組，右側可以是任何序列——字串、串列或元組。
例如，要將電子郵件地址分割成使用者名稱和網域，你可以寫：

```python
email = 'monty@python.org'
username, domain = email.split('@')
```

`split` 的回傳值是一個包含兩個元素的串列——第一個元素指派給 `username`，第二個指派給 `domain`。

```python
username, domain
```

左側的變數數量和右側的值數量必須相同——否則你會得到 `ValueError`。

```python
%%expect ValueError
a, b = 1, 2, 3
```

如果你想要交換兩個變數的值，元組指派很有用。
使用傳統的指派，你必須使用暫時變數，像這樣：

```python
temp = a
a = b
b = temp
```

這可以運作，但使用元組指派我們可以在不使用暫時變數的情況下做同樣的事情。

```python
a, b = b, a
```

這有效是因為右側的所有表達式都在任何指派之前被評估。

我們也可以在 `for` 陳述式中使用元組指派。
例如，要迴圈遍歷字典中的項目，我們可以使用 `items` 方法。

```python
d = {'one': 1, 'two': 2}

for item in d.items():
    key, value = item
    print(key, '->', value)
```

每次透過迴圈，`item` 被指派一個包含鍵和對應值的元組。

我們可以更簡潔地寫這個迴圈，像這樣：

```python
for key, value in d.items():
    print(key, '->', value)
```

每次透過迴圈，鍵和對應的值直接指派給 `key` 和 `value`。

## 元組作為回傳值

嚴格來說，函數只能回傳一個值，但如果該值是元組，效果與回傳多個值相同。
例如，如果你想要將兩個整數相除並計算商和餘數，計算 `x//y` 然後再計算 `x%y` 是低效的。最好同時計算它們。

內建函數 `divmod` 接受兩個參數並回傳包含兩個值的元組，商和餘數。

```python
divmod(7, 3)
```

我們可以使用元組指派將元組的元素儲存在兩個變數中。

```python
quotient, remainder = divmod(7, 3)
quotient
```

```python
remainder
```

這是一個回傳元組的函數例子。

```python
def min_max(t):
    return min(t), max(t)
```

`max` 和 `min` 是尋找序列中最大和最小元素的內建函數。
`min_max` 計算兩者並回傳包含兩個值的元組。

```python
min_max([2, 4, 1, 3])
```

我們可以像這樣將結果指派給變數：

```python
low, high = min_max([2, 4, 1, 3])
low, high
```

## 參數封裝

函數可以接受可變數量的參數。
以 `*` 運算子開頭的參數名稱會將參數**封裝**到元組中。
例如，以下函數接受任意數量的參數並計算它們的算術平均數——也就是它們的總和除以參數的數量。

```python
def mean(*args):
    return sum(args) / len(args)
```

參數可以是你喜歡的任何名稱，但 `args` 是慣例。
我們可以像這樣呼叫函數：

```python
mean(1, 2, 3)
```

如果你有一個值的序列，並且你想要將它們作為多個參數傳遞給函數，你可以使用 `*` 運算子來**解封裝**元組。
例如，`divmod` 正好接受兩個參數——如果你傳遞元組作為參數，你會得到錯誤。

```python
%%expect TypeError
t = (7, 3)
divmod(t)
```

即使元組包含兩個元素，它也算作單一參數。
但如果你解封裝元組，它會被視為兩個參數。

```python
divmod(*t)
```

如果你想要適應現有函數的行為，封裝和解封裝會很有用。
例如，這個函數接受任意數量的參數，移除最低和最高的，並計算其餘的平均數。

```python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
```

首先它使用 `min_max` 來尋找最低和最高的元素。
然後它將 `args` 轉換為串列，以便可以使用 `remove` 方法。
最後它解封裝串列，以便元素作為個別參數傳遞給 `mean`，而不是作為單一串列。

這是一個顯示效果的例子。

```python
mean(1, 2, 3, 10)
```

```python
trimmed_mean(1, 2, 3, 10)
```

這種「裁剪」平均數在一些主觀評分的運動中使用——如跳水和體操——以減少偏離其他評分的裁判的影響。

## Zip

元組對於迴圈遍歷兩個序列的元素並對對應元素執行操作很有用。
例如，假設兩個隊伍進行七場比賽的系列賽，我們將他們的分數記錄在兩個串列中，每個隊伍一個。

```python
scores1 = [1, 2, 4, 5, 1, 5, 2]
scores2 = [5, 5, 2, 2, 5, 2, 3]
```

讓我們看看每個隊伍贏了多少場比賽。
我們將使用 `zip`，這是一個內建函數，接受兩個或更多序列並回傳一個 **zip 物件**，之所以這樣稱呼是因為它將序列的元素配對，就像拉鍊的齒一樣。

```python
zip(scores1, scores2)
```

我們可以使用 zip 物件來成對迴圈遍歷序列中的值。

```python
for pair in zip(scores1, scores2):
     print(pair)
```

每次透過迴圈，`pair` 被指派一個分數的元組。
所以我們可以將分數指派給變數，並像這樣計算第一隊的勝利：

```python
wins = 0
for team1, team2 in zip(scores1, scores2):
    if team1 > team2:
        wins += 1
        
wins
```

可惜，第一隊只贏了三場比賽，輸了系列賽。

如果你有兩個串列並且想要一個配對的串列，你可以使用 `zip` 和 `list`。

```python
t = list(zip(scores1, scores2))
t
```

結果是一個元組的串列，所以我們可以像這樣取得最後一場比賽的結果：

```python
t[-1]
```

如果你有鍵的串列和值的串列，你可以使用 `zip` 和 `dict` 來建立字典。
例如，這是我們如何建立一個從每個字母對應到它在字母表中位置的字典。

```python
letters = 'abcdefghijklmnopqrstuvwxyz'
numbers = range(len(letters))
letter_map = dict(zip(letters, numbers))
```

現在我們可以查找一個字母並取得它在字母表中的索引。

```python
letter_map['a'], letter_map['z']
```

在這個對應中，`'a'` 的索引是 `0`，`'z'` 的索引是 `25`。

如果你需要迴圈遍歷序列的元素和它們的索引，你可以使用內建函數 `enumerate`。

```python
enumerate('abc')
```

結果是一個 **enumerate 物件**，它迴圈遍歷配對的序列，其中每個配對包含一個索引（從 0 開始）和來自給定序列的元素。

```python
for index, element in enumerate('abc'):
    print(index, element)
```

## 比較與排序

關係運算子適用於元組和其他序列。
例如，如果你對元組使用 `<` 運算子，它從每個序列的第一個元素開始比較。
如果它們相等，它會繼續到下一對元素，依此類推，直到找到不同的一對。

```python
(0, 1, 2) < (0, 3, 4)
```

後續元素不被考慮——即使它們真的很大。

```python
(0, 1, 2000000) < (0, 3, 4)
```

這種比較元組的方式對於排序元組串列或尋找最小值或最大值很有用。
作為例子，讓我們尋找單詞中最常見的字母。
在前一章中，我們撰寫了 `value_counts`，它接受字串並回傳一個從每個字母對應到它出現次數的字典。

```python
def value_counts(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

這是字串 `'banana'` 的結果。

```python
counter = value_counts('banana')
counter
```

只有三個項目，我們可以輕易看出最頻繁的字母是 `'a'`，出現三次。
但如果項目更多，自動排序會很有用。

我們可以像這樣從 `counter` 取得項目。

```python
items = counter.items()
items
```

結果是一個 `dict_items` 物件，其行為像元組的串列，所以我們可以像這樣排序它。

```python
sorted(items)
```

預設行為是使用每個元組的第一個元素來排序串列，並使用第二個元素來解決並列情況。

但是，要找到計數最高的項目，我們想要使用第二個元素來排序串列。
我們可以撰寫一個接受元組並回傳第二個元素的函數來做到這一點。

```python
def second_element(t):
    return t[1]
```

然後我們可以將該函數作為名為 `key` 的可選參數傳遞給 `sorted`，這表示應該使用此函數來計算每個項目的**排序鍵**。

```python
sorted_items = sorted(items, key=second_element)
sorted_items
```

排序鍵決定了串列中項目的順序。
計數最低的字母出現在第一位，計數最高的字母出現在最後。
所以我們可以像這樣找到最常見的字母。

```python
sorted_items[-1]
```

如果我們只想要最大值，我們不必排序串列。
我們可以使用 `max`，它也接受 `key` 作為可選參數。

```python
max(items, key=second_element)
```

要找到計數最低的字母，我們可以用同樣的方式使用 `min`。

## 反轉字典

假設你想要反轉字典，以便可以查找值並取得對應的鍵。
例如，如果你有一個從每個單詞對應到它出現次數的單詞計數器，你可以建立一個從整數對應到出現該次數的單詞的字典。

但有一個問題——字典中的鍵必須是唯一的，但值不需要。例如，在單詞計數器中，可能有許多單詞具有相同的計數。

所以反轉字典的一種方法是建立一個新字典，其中值是來自原始字典的鍵串列。
作為例子，讓我們計算 `parrot` 中的字母。

```python
d =  value_counts('parrot')
d
```

如果我們反轉這個字典，結果應該是 `{1: ['p', 'a', 'o', 't'], 2: ['r']}`，這表示出現一次的字母是 `'p'`、`'a'`、`'o'` 和 `'t'`，出現兩次的字母是 `'r'`。

以下函數接受字典並回傳其反轉作為新字典。

```python
def invert_dict(d):
    new = {}
    for key, value in d.items():
        if value not in new:
            new[value] = [key]
        else:
            new[value].append(key)
    return new
```

`for` 陳述式迴圈遍歷 `d` 中的鍵和值。
如果值還不在新字典中，它會被新增並與包含單一元素的串列關聯。
否則它會被附加到現有串列。

我們可以像這樣測試它：

```python
invert_dict(d)
```

我們得到了預期的結果。

這是我們看到的第一個例子，其中字典中的值是串列。

### 練習

在本章中我說元組可以用作字典中的鍵，因為它們是可雜湊的，而它們是可雜湊的因為它們是不可變的。
但這並不總是真的。

```python
# Solution

def shift_word(word, n):
    """Shift the letters of `word` by `n` places.
    
    >>> shift_word('cheer', 7)
    'jolly'
    >>> shift_word('melon', 16)
    'cubed'
    """
    t = []
    for letter in word:
        index = (letter_map[letter] + n) % 26
        t.append(letters[index])
    return ''.join(t)
```

```python
# Solution

def word_distance(word1, word2):
    """Computes the number of places where two word differ.

    >>> word_distance("hello", "hxllo")
    1
    >>> word_distance("ample", "apply")
    2
    >>> word_distance("kitten", "mutton")
    3
    """
    assert len(word1) == len(word2)

    count = 0
    for c1, c2 in zip(word1, word2):
        if c1 != c2:
            count += 1

    return count
```

撰寫一個程式來找到所有可以這樣簡化的單詞，然後找到最長的一個。

這個練習比大多數練習更有挑戰性，所以這裡有一些建議：

1.  你可能想要撰寫一個函數，它接受一個單詞並計算透過移除一個字母可以形成的所有單詞的串列。
    這些是單詞的「子項」。

2.  遞迴地，如果一個單詞的任何子項是可簡化的，那麼這個單詞就是可簡化的。作為基本情況，你可以認為空字串是可簡化的。

3.  我們一直使用的單詞串列不包含單個字母的單詞。所以你可能必須新增「I」和「a」。

4.  為了提高程式的效能，你可能想要備忘錄化已知可簡化的單詞。

```python
# Solution

def children(word):
    """Returns a list of all words that can be formed by removing one letter.

    word: string

    Returns: list of strings
    """
    res = []
    for i in range(len(word)):
        child = word[:i] + word[i+1:]
        if child in word_dict:
            res.append(child)
    return res
```

```python
# Solution

"""memo is a dictionary that maps from each word that is known
to be reducible to a list of its reducible children. 
It starts with the empty string."""

memo = {}
memo[''] = ['']


def reduce_word(word):
    """If word is reducible, returns a list of its reducible children.

    Also adds an entry to the memo dictionary.

    A string is reducible if it has at least one child that is 
    reducible.  The empty string is also reducible.

    word: string
    """
     # if have already checked this word, return the answer
    if word in memo:
        return memo[word]

    # check each of the children and make a list of the reducible ones
    res = []
    for child in children(word):
        if reduce_word(child):
            res.append(child)

    # memoize and return the result
    memo[word] = res
    return res
```

```python
# Solution

def print_trail(word):
    """Prints the sequence of words that reduces this word to the empty string.

    If there is more than one choice, it chooses the first.

    word: string
    """
    if len(word) == 0:
        return
    print(word, end=' ')
    t = reduce_word(word)
    print_trail(t[0])
```

```python
# Solution

def all_reducible():
    """Checks all words in the word_dict; returns a list of reducible ones.
    """
    res = []
    for word in word_dict:
        t = reduce_word(word)
        if len(t) > 0:
            res.append(word)
    return res
```

```python
# Solution

word_dict = {}
for word in word_list:
    word_dict[word] = 1

# have to add single letter words to the word list;
# also, the empty string is considered a word.
for word in ['a', 'i', '']:
    word_dict[word] = 1

# find the reducible words and sort by length
words = all_reducible()
sorted_words = sorted(words, key=len)

# print the longest words
for word in sorted_words[-10:]:
    print_trail(word)
    print('')
```
